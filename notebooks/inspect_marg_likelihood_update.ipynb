```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import pickle

from cmcrameri import cm
from pqcd.utils import nsat
```

```python
with open('../data/eos-extensions/eos_extensions_s-G-1p25-0p25_l-U-1-20_meancs2-G-0.3-0.3_pQCD-25-40.pickle.old', 'rb') as f:
    eos_extensions_old = pickle.load(f)

eos_extensions_old['n'] = (eos_extensions_old['p'] + eos_extensions_old['e'])/eos_extensions_old['mu']
```

```python
eos_extensions = pd.read_hdf(
    '../data/eos-extensions/eos_extensions_s-G-1p25-0p25_l-U-1-20_meancs2-G-0.3-0.3_pQCD-25-40.h5',
)
```

```python
eos_extensions['n'] = (eos_extensions['p'] + eos_extensions['e'])/eos_extensions['mu']
```

```python
N_samples = 1000

fig, axs = plt.subplots(1, 2, figsize=(12,4), sharex=True, sharey=True)

for ax in axs:
    ax.set_axisbelow(True)
    ax.grid(alpha=0.3)

for index, eos in eos_extensions_old.sample(N_samples).iterrows():
    axs[0].plot(eos.n[::-1]/nsat, eos.phi, c=cm.lipari(0.4), alpha=0.75, zorder=0.25, lw=1)

for index, eos in eos_extensions.sample(N_samples).iterrows():
    axs[1].plot(eos.n/nsat, eos.phi, c=cm.lipari(0.4), alpha=0.75, zorder=0.25, lw=1)

axs[0].set_title('Old EOS extensions')
axs[1].set_title('New EOS extensions')

for ax in axs:
    ax.set_xlabel(r'$n/n_\mathrm{sat}$')
axs[0].set_ylabel(r'$\phi$');
```

```python
nterm = 6*nsat
N_samples = 1000

fig, axs = plt.subplots(1, 2, figsize=(12,4), sharex=True, sharey=True)

for ax in axs:
    ax.set_axisbelow(True)
    ax.grid(alpha=0.3)

for index, eos in eos_extensions_old.sample(N_samples).iterrows():
    axs[0].plot(eos.e, eos.p, c=cm.lipari(0.4), alpha=0.75, zorder=0.25, lw=1)
    highlight = np.argmin(np.abs(eos.n[:-1] - nterm))
    axs[0].scatter(eos.e[highlight], eos.p[highlight], c='k', s=10, zorder=1)

for index, eos in eos_extensions.sample(N_samples).iterrows():
    axs[1].plot(eos.e, eos.p, c=cm.lipari(0.4), alpha=0.75, zorder=0.25, lw=1)
    highlight = np.argmin(np.abs(eos.n[1:] - nterm))
    axs[1].scatter(eos.e[highlight], eos.p[highlight], c='k', s=10, zorder=1)

axs[0].set_title('Old EOS extensions')
axs[1].set_title('New EOS extensions')

for ax in axs:
    ax.set_xlabel(r'$e$ [GeV/fm$^3$]')
axs[0].set_ylabel(r'$p$ [GeV/fm$^3$]');
```

```python
N_samples = 1000

fig, axs = plt.subplots(1, 2, figsize=(12,4), sharex=True, sharey=True)

for ax in axs:
    ax.set_axisbelow(True)
    ax.grid(alpha=0.3)

for index, eos in eos_extensions_old.sample(N_samples).iterrows():
    axs[0].plot(eos.e, eos.p, c=cm.lipari(0.4), alpha=0.75, zorder=0.25, lw=1)
    highlight = np.argmin(np.abs(eos.n[:-1] - nterm))
    axs[0].scatter(eos.e[highlight], eos.p[highlight], c='k', s=10, zorder=1)
    highlight = np.argmin(np.abs(eos.n[:-1] - 10*nsat))
    axs[0].scatter(eos.e[highlight], eos.p[highlight], c='C0', s=10, zorder=1)

for i, (index, eos) in enumerate(eos_extensions.sample(N_samples).iterrows()):
    if i == 0:
        label1 = 'EOS extensions'
        label2 = r'EOS extensions at 6 $n_\mathrm{sat}$'
        label3 = r'EOS extensions at 10 $n_\mathrm{sat}$'
    else:
        label1 = None
        label2 = None
        label3 = None
    axs[1].plot(eos.e, eos.p, c=cm.lipari(0.4), alpha=0.75, zorder=0.25, lw=1, label=label1)
    highlight = np.argmin(np.abs(eos.n[1:] - 6*nsat))
    axs[1].scatter(eos.e[highlight], eos.p[highlight], c='k', s=10, zorder=1, label=label2)
    highlight = np.argmin(np.abs(eos.n[1:] - 10*nsat))
    axs[1].scatter(eos.e[highlight], eos.p[highlight], c='C0', s=10, zorder=1, label=label3)

axs[0].set_title('Old EOS extensions')
axs[1].set_title('New EOS extensions')

axs[0].set_xlim(0)
axs[0].set_ylim(0)

axs[1].legend()

for ax in axs:
    ax.set_xlabel(r'$e$ [GeV/fm$^3$]')
axs[0].set_ylabel(r'$p$ [GeV/fm$^3$]');
```
